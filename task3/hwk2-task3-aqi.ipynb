```python
# Load in the packages
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 

# Load in the Data 
aqi_17 = pd.read_csv('data/daily_aqi_by_county_2017.zip', compression = 'zip')
aqi_18 = pd.read_csv('data/daily_aqi_by_county_2018.zip', compression = 'zip')


aqi = pd.concat([aqi_17, aqi_18]) # Combining the dataframes together

#Editing columns, making everything lowercase, adding underscores. 
aqi.columns = (aqi.columns
                  .str.lower()
                  .str.replace(' ','_')
                )


aqi_sb = aqi[aqi['county_name'] == 'Santa Barbara'] # Selecting data only from Santa Barbara, storing it as aqi_sb
aqi_sb = aqi_sb.drop( # Dropping state and county columns 
    columns = ['state_name', 'county_name', 'state_code', 'county_code'])


aqi_sb['date'] = pd.to_datetime(aqi_sb['date']) # Change to datetime
aqi_sb = aqi_sb.set_index('date') # Setting the index to the date column 

rolling_average = aqi_sb['aqi'].rolling('5D').mean() # Calculate AQI rolling average over 5 day averages 


aqi_sb['five_day_average'] = rolling_average # Save rolling average 



# Creating the line plot 
aqi_sb_plot = aqi_sb.plot(
    kind = 'line',
    title = 'Santa Barbara County Daily AQI and 5-Day Average, 2017–2018',
    xlabel = 'Date',
    ylabel ='Air Quality Index (AQI)',
    y = ['aqi', 'five_day_average'],
    color = ['hotpink', 'blue'],
    figsize = (12, 5),
    linewidth = 1.5
)

aqi_sb_plot # Display plot 
```

In the next cell we will calculate an average over a [rolling window](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.rolling.html) using the `rolling()` method for `pandas.Series`:

- `rolling()` is a lazy method, so we need to specify what we want to calculate over each window before it does something. 
- in this example we use the aggregator function `mean()` to calculate the average over each window
- the parameter '5D' indicates we want the window for our rolling average to be 5 days. 
- we get a `pandas.Series` as output

Store your answer in the `rolling_average` variable. Your result should be a `pandas.Series` indexed by `date`, with the 5-day averages as its values. 

```python
rolling_average = aqi_sb['aqi'].rolling('5D').mean() # Calculate AQI rolling average over 5 days
rolling_average # Display the dataframe 
```
